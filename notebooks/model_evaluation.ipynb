## 1. Problem Statement

The objective of this project is to build and evaluate machine learning classification models that predict whether a breast tumor is malignant or benign using numerical measurements from the Breast Cancer Wisconsin Diagnostic dataset.

Three models are compared: a Dummy Classifier as the baseline, Logistic Regression as a linear classification model, and Random Forest as an ensemble learning model.

Model evaluation is important because a high accuracy score alone does not guarantee that a model performs well on unseen data. Different evaluation metrics provide different insights into model performance, including how many positive cases are correctly identified and how many predictions are incorrect.

In this experiment, a **false negative** occurs when a malignant tumor is incorrectly classified as benign. This is an important error because the model fails to identify a malignant case. Therefore, recall for the malignant class is a particularly important metric to monitor.

The main goal is to compare the three models using multiple evaluation metrics, investigate their classification errors, and identify the model that performs best according to the chosen evaluation priorities.


## 2. Dataset Exploration

This project uses the Breast Cancer Wisconsin Diagnostic dataset provided by scikit-learn. The dataset contains 569 samples and 30 numerical features describing characteristics of cell nuclei derived from breast mass images.

The target variable represents two classes: malignant and benign. For this experiment, malignant cases are treated as the positive class to make the interpretation of precision, recall, and false negatives consistent.

Before training the models, the dataset is examined to understand its structure, feature names, target distribution, missing values, and class balance. These checks help identify potential data quality issues and provide an overview of the classification problem.

The dataset contains 212 malignant samples and 357 benign samples. This indicates a moderate class imbalance, with benign cases representing the majority class. A Dummy Classifier that always predicts the majority class can therefore achieve a baseline accuracy of approximately 63.16% without learning meaningful relationships between the features and the target.

Understanding the target distribution is important because accuracy can be misleading when one class is more common than another. For this reason, additional metrics such as precision, recall, F1-score, and ROC-AUC are used to evaluate the models.

The dataset is a well-established educational benchmark and is suitable for demonstrating classification and evaluation techniques. It should not be treated as a substitute for clinical testing or professional medical judgment.


## 3. Experimental Design

The dataset was divided into training and testing sets using an 80/20 train-test split. The training set was used to fit the models, while the test set was reserved for evaluating their performance on unseen samples.

A random seed of 42 was used to make the experiment reproducible. Stratified sampling was applied to preserve approximately the same proportion of malignant and benign cases in both sets. This helps ensure that the test set provides a more representative evaluation of the original target distribution.

All three models were trained and evaluated using the same data split to ensure a fair comparison.

The experiment includes three models:

1. **Dummy Classifier:** Establishes a baseline by predicting the most frequent class.
2. **Logistic Regression:** Provides a simple, interpretable classification approach. StandardScaler is applied through a pipeline so that scaling parameters are learned from the training data only.
3. **Random Forest:** Uses an ensemble of decision trees to capture potentially nonlinear relationships between features.

The test set was kept separate from model training and model selection. This reduces the risk of data leakage and provides a more reliable estimate of how the models perform on unseen data.

Performance was evaluated using accuracy, precision, recall, F1-score, ROC-AUC, and confusion matrices. Malignant cases were defined as the positive class for the precision, recall, and F1-score calculations.

The same split and evaluation procedure were used for all three models, and no hyperparameter tuning was performed against the test set.


## 4. Train and Compare Models

Three classification models were trained and compared using the same training and testing datasets.

The Dummy Classifier served as a baseline to establish the performance of a simple majority-class prediction strategy. Logistic Regression was used as a linear model, while Random Forest was selected to evaluate whether an ensemble of decision trees could improve classification performance.

The results show a substantial improvement when using learned models instead of the baseline.

The Dummy Classifier achieved an accuracy of 63.16%, but its precision, recall, and F1-score for the malignant class were all zero. This happened because the classifier predicted every sample as benign and failed to identify any malignant cases.

Logistic Regression achieved an accuracy of 96.49%, a precision of 97.50%, a recall of 92.86%, and an F1-score of 95.12%.

Random Forest achieved an accuracy of 97.37%, a precision of 100.00%, a recall of 92.86%, and an F1-score of 96.30%.

Random Forest performed best overall among the three models based on the observed test results. It achieved the highest accuracy and F1-score while maintaining the same recall as Logistic Regression. Its ROC-AUC score was also slightly higher.

These results suggest that both learned models substantially outperform the baseline on this test split. However, performance on a single split does not guarantee that the same ranking will hold on other datasets or future samples.


## 5. Results Visualization and Comparison

The evaluation metrics were organized into a comparison table to make it easier to identify differences between the models.

| Model               | Accuracy | Precision | Recall | F1-score | ROC-AUC |
| ------------------- | -------: | --------: | -----: | -------: | ------: |
| Random Forest       |   97.37% |   100.00% | 92.86% |   96.30% |  99.77% |
| Logistic Regression |   96.49% |    97.50% | 92.86% |   95.12% |  99.60% |
| Dummy Baseline      |   63.16% |     0.00% |  0.00% |    0.00% |  50.00% |

**Accuracy** measures the proportion of all predictions that were correct. Random Forest achieved the highest accuracy at 97.37%.

**Precision** measures the proportion of predicted malignant cases that were actually malignant. Random Forest achieved 100% precision on this test set, meaning none of its predicted malignant cases were false positives.

**Recall** measures the proportion of actual malignant cases correctly identified. Both Random Forest and Logistic Regression achieved 92.86% recall, meaning both identified the same proportion of malignant cases on this test split.

**F1-score** combines precision and recall into a single metric. Random Forest achieved the highest F1-score of 96.30%.

**ROC-AUC** measures how well a model ranks positive cases above negative cases across different classification thresholds. Both learned models achieved ROC-AUC scores above 0.99 on this test set.

The confusion matrices provide additional information about correct predictions and classification errors. They are particularly useful for examining false negatives, which represent malignant cases incorrectly classified as benign.

Overall, Random Forest produced the strongest combination of results in this experiment. However, its perfect precision should be interpreted as an observation from this particular test set rather than a guarantee of perfect precision in future predictions.


## 6. Error Analysis

Error analysis was used to understand the types of mistakes made by each model rather than relying only on aggregate evaluation metrics.

The Dummy Classifier failed to identify any malignant cases because it always predicted the majority class, benign. Its malignant-class recall was therefore zero. This demonstrates why a baseline model can achieve seemingly reasonable accuracy while being unsuitable for a classification task where both classes matter.

Logistic Regression achieved 92.86% recall and 97.50% precision. This means that it identified most malignant cases, although some malignant cases were missed and some benign cases were incorrectly classified as malignant.

Random Forest achieved the same recall of 92.86% but higher precision of 100%. On this test set, it correctly identified the same proportion of malignant cases as Logistic Regression without producing any false-positive malignant predictions.

The remaining false negatives are especially important because they represent malignant cases classified as benign. Such errors could have serious consequences in a real medical setting, which is why recall and the confusion matrix are important in this experiment.

Individual misclassified samples can be inspected using the generated error analysis file. Comparing the errors made by different models can help identify whether they struggle with the same samples or make different types of mistakes.

The observed results show that Random Forest made fewer overall classification errors than Logistic Regression on this split. However, identifying the underlying reasons for individual mistakes requires further investigation of the relevant feature values and model behavior.

This analysis is limited to the available test samples. A larger independent test set and cross-validation would provide additional evidence about model stability and generalization.


## 7. Conclusion

This project compared a Dummy Classifier, Logistic Regression, and Random Forest on the Breast Cancer Wisconsin Diagnostic dataset.

The results demonstrated that both learned models substantially outperformed the baseline. The Dummy Classifier achieved 63.16% accuracy but failed to identify any malignant cases, showing that accuracy alone is not sufficient to evaluate classification performance.

Logistic Regression achieved 96.49% accuracy and a 95.12% F1-score. Random Forest achieved the best overall results on the test split, with 97.37% accuracy, 100% precision, 92.86% recall, a 96.30% F1-score, and a ROC-AUC of 99.77%.

Random Forest was the strongest model in this experiment according to the observed accuracy, precision, F1-score, and ROC-AUC. However, its recall was identical to that of Logistic Regression, indicating that both models identified the same proportion of malignant cases on this split.

The main lesson from this experiment is that a model should not be judged by accuracy alone. Comparing multiple metrics and investigating false positives and false negatives provides a more complete understanding of its performance.

Future improvements include using stratified cross-validation, tuning hyperparameters using only the training data, investigating feature importance, evaluating probability calibration, and examining alternative decision thresholds. The final model should then be evaluated on an untouched test set.
